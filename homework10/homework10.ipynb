```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from astropy import units as u
from astropy import constants as const
```

```python
df2 = pd.read_csv("solar_system.csv")
```

```python
print(df2.shape)
```

```python
df2.head()
```

```python
df = df2.set_index("Attribute").T
#sets "attribute" as the index and then transposes with .T, flipping the columns and the rows
df.index.name = "Planet"
#changes 'attribure" to "planet"
df.reset_index(inplace=True)
#keeps everything in place and makes planets a regular column
df.columns.name = None
#removes any names from the column axis
```

```python
df.shape
```

```python
#(20,11) before transposing and (10,21) after transposing
#the new shape is different becomes the columns and rows have been flipped. Also, the "attribute" from the rows got moved into columns as "planets"
print(df.columns)
```

```python
# columns with units: mass, diameter, density, gravity, escape velocity, rotation period, length of day, distance from sun, perihelion, aphelion, orbital period, orbital velocity, orbital inclination, orbital eccentricity, obliquity to orbit, mean temperature, surface pressure
# columns without units: planet, number of moons, ring system?, global magnetic field?
print(df)
```

```python
df.dtypes
```

```python
for name in df.columns:
    print(name)
    print(df[name].apply(type))
```

```python
#the types are all strings
#this is a problem becuase the computer will not interpret them as numbers and you therefore cannot perform mathmatical operations on them
```

```python
skip = ["Planet", "Ring System?", "Global Magnetic Field?"]
for col in df.columns:
    if col not in skip:
        df[col] = pd.to_numeric(df[col], errors="coerce")
```

```python
df.dtypes
```

```python
# float64, int64, and object
#float64 are numbers with decimals with an 64-bit processor, while int64 are integer numbers with an 64-bit processor, and object is a mix of numbers and strings
#Planet, ring system, and global magnetic field because I excluded them from the function converting all strings to numbers
```

```python
def attach_units (units, old_col, new_col):
    new_values = []
    for val in df[old_col]:
        newvals = val * units
        new_values.append(newvals)
    df[old_col] = new_values
    df.rename(columns={old_col: new_col}, inplace=True)

unit = 1e6 * u.km
col = "Distance from Sun (10^6 km)"
newcol = "Distance from Sun (km)"

attach_units(unit, col, newcol)
```

```python
attach_units(unit, "Perihelion (10^6 km)", "Perihelion (km)")
```

```python
attach_units(unit, "Aphelion (10^6 km)", "Aphelion (km)")
```

```python
print(df["Distance from Sun (km)"])
```

```python
df["Semi-major Axis (km)"] = (df["Perihelion (km)"] + df["Aphelion (km)"]) / 2
print(df.columns)
print(df["Semi-major Axis (km)"])
```

```python
unit2 = u.year / 365.2422 
attach_units(unit2, "Orbital Period (days)", "Orbital Period (years)")
```

```python
unit3 = u.au / (149597870.70000002 * u.km)
attach_units(unit3, "Distance from Sun (km)", "Distance from Sun (AU)")
```
