```python
import csv
import numpy as np
import pandas as pd
from scipy.signal import find_peaks
from scipy.optimize import curve_fit
import matplotlib.pyplot as plt
```

```python
#estimating muon lifespan based on the second dataset
```

```python

# 1. Load CSV manually
rows = []
with open("MD20231005_1151.csv") as i:
    reader = csv.reader(i)
    for r in reader:
        rows.append(r)

print(len(rows))

df = pd.DataFrame(rows)
print(df.head())
```

```python
#collecting data times - different from first dataset, since this one shows pulses
decay_times = [] 

#defining variables for the scipy.signal find_peaks function
interval = 1.25   # microseconds between each sample (given by experiment data)
average_height = -0.013   # average level of dip (noise)
buffer_between_peaks = 3  # spaces data out so it doesn't detect too many peaks 


#cleaning up data to make sure every index is integer and every number is a float
for row in rows:
    if row[1].isdigit():
        int(row[1]) 
    else:
        continue
    try:
        wave = np.array([float(x) for x in row[2:] if x != ""])
        #makes the numbers floats and collects them in wave list
    except:
        continue
    #checks for peaks
    if len(wave) < 3:
        continue
    # makes peaks positive
    make_positive = -wave
    
#uses find_peaks scipy function to detect where the peaks happen (muon decayed and the electron appeared)
    peaks, props = find_peaks(
        make_positive,
        height=-average_height, 
        distance=buffer_between_peaks
    )

    if len(peaks) >= 2:
        t1 = peaks[0] * interval #first peak shows where the muon stops
        t2 = peaks[1] * interval #second peak shows where the decay electron appears

        decay_time = t2 - t1
        decay_times.append(decay_time)

#number of detected decays
print(f"Detected decay events: {len(decay_times)}")

#histogram showing distributing of decay times
plt.hist(decay_times, bins=25, color='green')
plt.xlabel("Decay time (μs)")
plt.ylabel("Frequency of Occurence")
plt.title("Muon Decay Time Distribution")
plt.show()
```

```python
#log scale to help condense the data for easier data analysis
plt.hist(decay_times, bins=25, log = True, color='green')
plt.xlabel("Decay time (μs)")
plt.ylabel("Frequency of Occurence")
plt.title("Muon Decay Time Distribution")
#plt.show()
```

```python
# exponential fit based on the exponential decay law
def exponential_fit(t, C, tau):
    return C * np.exp(-t/tau)

count, bins = np.histogram(decay_times, bins = 25)
#take the center of each bin for exponential fitting
centers = 0.5*(bins[1:] + bins[:-1])

popt, pcov = curve_fit(exponential_fit, centers, count, p0 = [max(count), 2.0])
C_fit, tau_fit = popt

print(f"Estimated muon lifetime {tau_fit:.5f} microseconds")
```

```python
#plotting the histogram together with the line of best fit

plt.hist(decay_times, bins = 25, log = True, color = "green")
plt.plot(centers, exponential_fit(centers, *popt), 'r-', label=f"Fit τ={tau_fit:.3f} μs")
plt.yscale("log")
plt.xlabel("Decay time [μs]")
plt.ylabel("Frequency of Occurence (log)")
plt.title("Muon Decay Time Distribution")
plt.legend()
plt.show()
```
